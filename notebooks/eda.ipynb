```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/kc_house_data.csv")

df.head()
```

```python
df.shape
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df.describe()
```

```python
# check for missing values
df.isnull().sum()
```

```python
# price distribution

plt.figure()

sns.histplot(
    df["price"],
    bins=50
)

plt.title("House Price Distribution")

plt.show()
```

```python
# checking for outliers

plt.figure()

sns.boxplot(
    x=df["price"]
)

plt.title("Price Outliers")

plt.show()
```

```python
# geographic distribution of price

plt.figure(figsize=(8,8))

plt.scatter(
    df["long"],
    df["lat"],
    c=df["price"],
    cmap="viridis",
    s=5
)

plt.colorbar(label="Price")

plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.title("Spatial Distribution of House Prices")

plt.show()
```
